```python
data = [
    ['blue','medium',0],
    ['green','small',1],
    ['blue','small',0],
    ['red','large',1],
    ['blue','medium',0]
]
```

```python
def find_s_algorithm(data):
  hypothesis=[None,None]
  for example in data:
    if example[-1]==1:
      for i in range(len(hypothesis)):
        if hypothesis[i] is None:
          hypothesis[i]= example[i]
        elif hypothesis[i]!= example[i]:
          hypothesis[i]="?"
    return hypothesis

```

```python
specific_hypothsis = find_s_algorithm(data)
```
